```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import pearsonr

# =========================
# Datos finales (promedios por sujeto)
# =========================
subjects = ['01','02','03','04','05','06','07','08','09','10']

ground_truth = np.array([
    16.66, 16.67, 16.33, 16.67, 16.33,
    17.00, 20.00, 22.33, 19.33, 16.67
])

watchy_estimated = np.array([
    15.33, 17.33, 18.00, 16.00, 16.00,
    15.33, 14.67, 16.00, 15.33, 16.67
])

# =========================
# Correlación
# =========================
correlation, p_value = pearsonr(ground_truth, watchy_estimated)

# =========================
# Figura
# =========================
plt.figure(figsize=(10, 8))

# Scatter
plt.scatter(
    ground_truth,
    watchy_estimated,
    s=150,
    alpha=0.7,
    color='#2E86AB',
    edgecolors='black',
    linewidth=1.5
)

# =========================
# Línea de regresión
# =========================
min_val = min(ground_truth.min(), watchy_estimated.min()) - 1
max_val = max(ground_truth.max(), watchy_estimated.max()) + 1

z = np.polyfit(ground_truth, watchy_estimated, 1)
p = np.poly1d(z)

x_line = np.linspace(min_val, max_val, 100)
plt.plot(
    x_line,
    p(x_line),
    color='green',
    linewidth=2,
    alpha=0.8,
    label=f'Linear fit (r = {correlation:.3f})'
)

# =========================
# Etiquetas por sujeto
# =========================
for i, subj in enumerate(subjects):
    plt.annotate(
        f'S{subj}',
        (ground_truth[i], watchy_estimated[i]),
        xytext=(6, 6),
        textcoords='offset points',
        fontsize=10,
        alpha=0.8
    )

# =========================
# Estética
# =========================
plt.xlabel('Ground-truth steps (average)', fontsize=13, fontweight='bold')
plt.ylabel('Watchy-estimated steps (average)', fontsize=13, fontweight='bold')
plt.title(
    'Step-count validation: Ground-truth vs Watchy-estimated steps',
    fontsize=14,
    fontweight='bold',
    pad=20
)

plt.legend(fontsize=11, loc='upper left')
plt.grid(True, alpha=0.3, linestyle='--')

# =========================
# Caja con métricas
# =========================
textstr = (
    f'Pearson r = {correlation:.3f}\n'
    f'p-value = {p_value:.4f}'
)

props = dict(boxstyle='round', facecolor='wheat', alpha=0.5)

plt.text(
    0.97,
    0.05,
    textstr,
    transform=plt.gca().transAxes,
    fontsize=11,
    verticalalignment='bottom',
    horizontalalignment='right',
    bbox=props
)

plt.tight_layout()

# =========================
# Guardar figura
# =========================
plt.savefig(
    'watchy_steps_validation.png',
    dpi=300,
    bbox_inches='tight'
)

print("✔ Validation graph saved as 'watchy_steps_validation.png'")
print(f"Pearson r = {correlation:.3f}")
print(f"P-value = {p_value:.4f}")

plt.close()
```

```python
# -*- coding: utf-8 -*-
"""
Validation of RPM measurements: Watchy vs Biaxial Oscillatory Table
Correlation plot and validation metrics (publication-ready)
"""

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from scipy import stats

# =========================
# Validation data
# Y-axis: Ground-truth RPM (biaxial table)
# X-axis: Watchy-estimated RPM
# =========================
rpm_mesa = np.array([25, 35, 50, 55, 60, 65, 70, 75])
rpm_watchy = np.array([25, 35, 50, 50, 60, 60, 65, 70])

# =========================
# Validation metrics
# =========================
r2 = r2_score(rpm_mesa, rpm_watchy)
mae = mean_absolute_error(rpm_mesa, rpm_watchy)
rmse = np.sqrt(mean_squared_error(rpm_mesa, rpm_watchy))

# Linear regression
slope, intercept, r_value, p_value, std_err = stats.linregress(
    rpm_watchy, rpm_mesa
)

# =========================
# Figure
# =========================
plt.figure(figsize=(10, 8))

# Scatter plot
plt.scatter(
    rpm_watchy,
    rpm_mesa,
    s=130,
    color='#2E86AB',
    alpha=0.7,
    edgecolors='black',
    linewidth=1.5,
    label='Measurements'
)

# =========================
# Regression line
# =========================
min_val = min(rpm_watchy.min(), rpm_mesa.min()) - 5
max_val = max(rpm_watchy.max(), rpm_mesa.max()) + 5

x_reg = np.linspace(min_val, max_val, 100)
y_reg = slope * x_reg + intercept

plt.plot(
    x_reg,
    y_reg,
    color='green',
    linewidth=2,
    alpha=0.8,
    label=f'Linear fit (R² = {r2:.3f})'
)

# =========================
# Point annotations
# =========================
for i in range(len(rpm_mesa)):
    plt.annotate(
        f'{rpm_watchy[i]} / {rpm_mesa[i]}',
        (rpm_watchy[i], rpm_mesa[i]),
        textcoords="offset points",
        xytext=(6, 6),
        fontsize=9,
        alpha=0.8
    )

# =========================
# Axis labels and title
# =========================
plt.xlabel('Watchy-estimated RPM', fontsize=13, fontweight='bold')
plt.ylabel('Ground-truth RPM (biaxial table)', fontsize=13, fontweight='bold')
plt.title(
    'RPM measurement validation: Watchy vs biaxial oscillatory table',
    fontsize=14,
    fontweight='bold',
    pad=20
)

# =========================
# Grid & legend
# =========================
plt.grid(True, alpha=0.3, linestyle='--')
plt.legend(loc='upper left', fontsize=11)

# =========================
# Metrics textbox
# =========================
textstr = (
    f'R² = {r2:.3f}\n'
    f'MAE = {mae:.2f} RPM\n'
    f'RMSE = {rmse:.2f} RPM\n'
    f'n = {len(rpm_mesa)}'
)

props = dict(boxstyle='round', facecolor='wheat', alpha=0.6)

plt.text(
    0.97,
    0.05,
    textstr,
    transform=plt.gca().transAxes,
    fontsize=11,
    verticalalignment='bottom',
    horizontalalignment='right',
    bbox=props
)

# =========================
# Axis limits
# =========================
plt.xlim(min_val, max_val)
plt.ylim(min_val, max_val)

plt.tight_layout()

# =========================
# Save figure
# =========================
output_file = 'watchy_rpm_validation.png'
plt.savefig(output_file, dpi=300, bbox_inches='tight')
plt.close()

print(f"✔ RPM validation figure saved as '{output_file}'")
print(f"R² = {r2:.3f}")
print(f"MAE = {mae:.2f} RPM")
print(f"RMSE = {rmse:.2f} RPM")
print(f"p-value = {p_value:.4e}")
```
